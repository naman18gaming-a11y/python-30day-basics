```python
import kagglehub
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

path = kagglehub.dataset_download("mukuldeshantri/ecommerce-fashion-dataset")
print("Dataset downloaded to:", path)

csv_path = f"{path}/ecommerce-fashion.csv"
csv_path = os.path.join(path, "ecommerce-fashion.csv")
if not os.path.exists(csv_path):
    for root, dirs, files in os.walk(path):
        for name in files:
            if name.lower().endswith(".csv"):
                csv_path = os.path.join(root, name)
                break
        else:
            continue
        break
    else:
        raise FileNotFoundError(f"No CSV file found under {path}")

df = pd.read_csv(csv_path)

```

```python
df['BrandName'] = df['BrandName'].replace('Nan','pd.NA')
```

```python
df['Deatils']=df['Deatils'].replace('Nan','pd.NA')
```

```python
df['Sizes']=df['Sizes'].replace('Nan','pd.NA')
df['MRP']=df['MRP'].replace('Nan','pd.NA')
df['SellPrice']=df['SellPrice'].replace('Nan','pd.NA')
df['Discount']=df['Discount'].replace('Nan','pd.NA')
df['Category']=df['Category'].replace('Nan','pd.NA')
```

```python
df['color'] = df['Deatils'].str.split('-').str.get(-1).str.strip()
```

```python
df["Deatils"] = df["Deatils"].str.replace(r"-\s*\w+$", "", regex=True).str.strip()
```

```python
# rename the misspelled column then apply the regex on the corrected name
df.rename(columns={'Deatils': 'Details'}, inplace=True)
df["Details"] = df["Details"].str.replace(r"-\s*[A-Za-z ]+$", "", regex=True).str.strip()
```

```python
# Define mask to filter rows where Details is not null and contains the pattern
mask = df["Details"].notna() & df["Details"].str.contains(r"-\s*[A-Za-z ]+$", regex=True, na=False)

# Apply the replacement only to rows matching the mask
df.loc[mask, "Details"] = df.loc[mask, "Details"].str.replace(r"-\s*[A-Za-z ]+$", "", regex=True).str.strip()
```

```python
color_counts =df['color'].value_counts()
print(color_counts)
```

```python
if 'ProductName' not in df.columns:
    df['ProductName'] = df['Details']

df['ProductName'] = df['ProductName'].str.lower()
df['color'] = df['color'].str.lower()
df['BrandName'] = df['BrandName'].astype(str).str.lower()

df['BrandName'] = df['BrandName'].str.lower()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a clean visual style for all plots
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['SellPrice'], bins=50, kde=True, color='purple')
plt.title('Distribution of Product Selling Prices', fontsize=14, fontweight='bold')
plt.xlabel('Selling Price', fontsize=12)
plt.ylabel('Number of Products', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.regplot(data=df, x='MRP', y='SellPrice', scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
plt.title('MRP vs. Selling Price Relationship', fontsize=14, fontweight='bold')
plt.xlabel('MRP', fontsize=12)
plt.ylabel('Sell Price', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
# Group by category and find average discount
cat_discount = df.groupby('Category')['Discount'].mean().reset_index()
cat_discount = cat_discount.sort_values(by='Discount', ascending=False)

sns.barplot(data=cat_discount, x='Discount', y='Category', palette='viridis')
plt.title('Average Discount Percentage by Category', fontsize=14, fontweight='bold')
plt.xlabel('Average Discount (%)', fontsize=12)
plt.ylabel('Category', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
top_brands = df['BrandName'].value_counts().head(10).reset_index()
top_brands.columns = ['BrandName', 'Count']

sns.barplot(data=top_brands, x='Count', y='BrandName', palette='crest')
plt.title('Top 10 Brands by Product Variety', fontsize=14, fontweight='bold')
plt.xlabel('Number of Products', fontsize=12)
plt.ylabel('Brand Name', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8, 8))
category_counts = df['Category'].value_counts().head(8) # Top 8 categories for clarity

plt.pie(category_counts, labels=category_counts.index, autopct='%1.1f%%', 
        startangle=140, colors=sns.color_palette('pastel'), 
        wedgeprops=dict(width=0.4, edgecolor='white')) # width=0.4 makes it a donut chart
plt.title('Product Distribution Across Categories', fontsize=14, fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
# Filter for top 8 brands to keep the chart clean and readable
top_8_brands = df['BrandName'].value_counts().head(8).index
filtered_df = df[df['BrandName'].isin(top_8_brands)]

sns.boxplot(data=filtered_df, x='SellPrice', y='BrandName', palette='Set2')
plt.title('Price Range Distribution for Top 8 Brands', fontsize=14, fontweight='bold')
plt.xlabel('Selling Price', fontsize=12)
plt.ylabel('Brand Name', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
top_colors = df['color'].value_counts().head(10).reset_index()
top_colors.columns = ['color', 'Count']

sns.barplot(data=top_colors, x='Count', y='color', palette='magma')
plt.title('Top 10 Most Common Colors in Dataset', fontsize=14, fontweight='bold')
plt.xlabel('Number of Products', fontsize=12)
plt.ylabel('Color', fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(12, 8))
# Get top 8 categories and top 8 colors to make the heatmap readable
top_categories = df['Category'].value_counts().head(8).index
top_colors_list = df['color'].value_counts().head(8).index

heatmap_data = df[df['Category'].isin(top_categories) & df['color'].isin(top_colors_list)]
pivot_table = heatmap_data.pivot_table(index='color', columns='Category', aggfunc='size', fill_value=0)

sns.heatmap(pivot_table, annot=True, fmt='d', cmap='Blues', linewidths=.5)
plt.title('Heatmap of Colors Across Top Categories', fontsize=14, fontweight='bold')
plt.xlabel('Category', fontsize=12)
plt.ylabel('Color', fontsize=12)
plt.show()
```
