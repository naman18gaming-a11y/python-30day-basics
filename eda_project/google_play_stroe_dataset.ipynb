```python
import os
import pandas as pd
path = kagglehub.dataset_download("lava18/google-play-store-apps")

print("Path to dataset files:", path)
file_path = os.path.join(path, 'app.csv')
file_path = os.path.join(path, 'googleplaystore.csv')
df = pd.read_csv(file_path)
print(df.shape)
```

```python
sns.set_style("whitegrid")
plt.figure(figsize=(12, 6))
sns.barplot(
    x=df['Category'].value_counts().index,
    y=df['Category'].value_counts().values
)

plt.xlabel("Category", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.title("Count of Apps by Category", fontsize=15)

plt.xticks(rotation=45, ha="right")  # rotate labels
plt.show()
```

```python
sns.set_style("whitegrid")
plt.figure(figsize=(12, 6))
sns.barplot(
    x=df['Category'].value_counts().index,
    y=df.groupby('Category')['Rating'].mean().values
)
plt.xlabel("Category", fontsize=12)
plt.ylabel("Ratings", fontsize=12)
plt.title("Ratings of Apps by Category", fontsize=15)
plt.xticks(rotation=45, ha="right")  # rotate labels
plt.show()
```

```python


# Count apps per genre
genre_counts = df['Genres'].value_counts()

# Show top 15 genres
top_genres = genre_counts.head(15)

plt.figure(figsize=(12,6))
sns.barplot(x=top_genres.index, y=top_genres.values, palette="mako")

plt.xlabel("Genres", fontsize=12)
plt.ylabel("Count of Apps", fontsize=12)
plt.title("Top 15 Genres of Apps", fontsize=15)

plt.xticks(rotation=45, ha="right")
plt.show()
```
