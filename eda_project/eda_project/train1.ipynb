```python
df['Group_size'] = df['Ticket'].map(df['Ticket'].value_counts())
print(df[['Name', 'Ticket', 'Group_size']].head())
```

```python
# 1. Create the Group_Size column
df['Group_size'] = df['Ticket'].map(df['Ticket'].value_counts())

# 2. Print the total count of people in each group category
print("Total Passenger Count by Ticket Group Size:")
print(df['Group_size'].value_counts().sort_index())
```

```python
print(df['Fare'].dtype)
```

```python
df['Fare']= df['Fare'].round(2)
print(df[['Name','Fare']].head())
```

```python
print(df['Cabin'].isnull().sum())
```

```python
df['Deck']= df['Cabin'].astype(str).str.strip().str[0]
print("Unique Deck values:", df['Deck'].unique())
```

```python
# Replace 'n' (the missing values) with 'Unknown'
df['Deck'] = df['Deck'].replace('n', 'Unknown')

# Print a clean value count of our newly engineered Deck column
print(df['Deck'].value_counts())
```

```python
# If missing, count is 0. If present, split by space and count how many cabins they had.
df['Cabin_Count'] = df['Cabin'].apply(lambda x: len(str(x).split()) if pd.notnull(x) else 0)

# Check the passengers who bought the most cabins
print(df[['Name', 'Cabin', 'Cabin_Count']].sort_values(by='Cabin_Count', ascending=False).head(5))
```

```python
deck_map = {label: idx for idx, label in enumerate(sorted(df['Deck'].dropna().unique()))}
df['Deck'] = df['Deck'].map(deck_map).astype('Int64')
print(deck_map)
print(df[['Deck']].head())
```

```python

plt.Figure(figsize=(6,4))
ax = sns.countplot(
    x ='Survived',
    data = df,
    palette= 'Set2',
)
ax.set_xticklabels(['Perished (0)', 'Survived (1)'])
plt.title("overall ")
plt.ylabel("number of passanger")
plt.show()
```

```python
plt.figure(figsize=(8,5))
ax =sns.countplot(x='Group_size',data=df,palette="magma")
plt.title("Ticket Group Size Distribution")
plt.xlabel("ticket group size")
plt.ylabel("the count of passanger")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(y='Group_size',x='Survived',data=df,errorbar=None,palette='viridis')
plt.xlabel('size of the group')
plt.ylabel('survival rate')
plt.title('Survival Rate by Ticket Group Size')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='Survived', y='Fare', data=df, palette='pastel')
plt.xticks([0, 1], ['Perished (0)', 'Survived (1)'])
plt.title('Ticket Fare Distribution by Survival Status')
plt.ylim(0, 300) # Limits view to ignore extreme $500 outliers for better scale
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='Pclass', y='Survived', hue='Sex', data=df, errorbar=None, palette='muted')
plt.title('Survival Rate by Passenger Class and Gender')
plt.xlabel('Passenger Class (1st, 2nd, 3rd)')
plt.ylabel('Survival Rate')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='Age', hue='Survived', multiple='stack', kde=True, palette='coolwarm', bins=30)
plt.title('Age Distribution of Passengers (Stacked by Survival Status)')
plt.xlabel('Age')
plt.ylabel('Passenger Count')
plt.show()
```
