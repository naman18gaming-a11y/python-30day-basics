```python
import kagglehub
import os
import pandas as pd
import glob
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("troubador/2026-world-cup-player-statistics")

print("Path to dataset files:", path)
```

```python
files = glob.glob(os.path.join(path, "*"))
print(files)
```

```python
df = pd.read_csv(files[0])
df.shape
```

```python
df.describe()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.rename(columns={
    "Rk": "Rank",
    "Player": "Player_Name",
    "Pos": "Position",
    "Squad": "Team",
    "Age": "Age",
    "Born": "Birth_Year",
    "MP": "Matches_Played",
    "Starts": "Matches_Started",
    "Min": "Minutes_Played",
    "90s": "Ninety_Minute_Equivalents",
    "Gls": "Goals",
    "Ast": "Assists",
    "G+A": "Goals_Assists",
    "G-PK": "Non_Penalty_Goals",
    "PK": "Penalty_Goals",
    "PKatt": "Penalty_Attempts",
    "CrdY": "Yellow_Cards",
    "CrdR": "Red_Cards",
    "PlayerID": "Player_ID",
    "AstPer90": "Assists_Per_90",
    "G+APer90": "Goal_Contributions_Per_90",
    "GlsPer90": "Goals_Per_90",
    "G-PKPer90": "Non_Penalty_Goals_Per_90",
    "G+A-PKPer90": "Non_Penalty_Goal_Contributions_Per_90",
    "SecondPos": "Secondary_Position"
}, inplace=True)
```

```python
df.columns
```

```python
df.drop(columns=["Secondary_Position"], inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())    
```

```python
df.duplicated().sum()
```

```python
top10_players = df.nlargest(10, 'Goals')
print(top10_players[['Player_Name', 'Team', 'Goals']])
```

```python
top10_assists_players = df.nlargest(10, 'Assists')
print(top10_assists_players[['Player_Name', 'Team', 'Assists']])
```

```python
top10_players_Yellow_Cards = top10_players.Yellow_Cards.fillna(0, inplace=True)
top10_players_Yellow_Cards = top10_players[['Player_Name', 'Team', 'Yellow_Cards']].fillna(0)
print(top10_players_Yellow_Cards)
```

```python
top10_ed_card = df.nlargest(10,'Red_Cards')
print(top10_ed_card[['Player_Name', 'Team', 'Red_Cards']])
```

```python
sns.barplot(data=df.nlargest(10, 'Goals'), x='Goals', y='Player_Name')
plt.title("Top 10 Goal Scorers")
plt.xlabel("goals")
plt.ylabel("Playername")
```

```python
sns.barplot(data=df.nlargest(10, 'Assists'), x='Assists', y='Player_Name')
plt.title("Top 10 Assists") 
plt.xlabel("Assists")
plt.ylabel("Playername")
```

```python
plt.figure(figsize=(12,8))

corr = df.select_dtypes(include="number").corr()
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
```

```python
sns.scatterplot(
    data=df,
    x="Goals",
    y="Assists",
    hue="Position"
)

plt.title("Goals vs Assists")
plt.show()
```

```python
sns.boxplot(
    data=df,
    x="Position",
    y="Goals"
)

plt.title("Goals by Position")
plt.show()
```

```python
sns.histplot(
    data=df,
    x="Age",
    bins=15,
    kde=True
)

plt.title("Age Distribution")
plt.show()
```

```python
team_goals = df.groupby("Team")["Goals"].sum().sort_values(ascending=False).head(10)

team_goals.plot(kind="bar")

plt.title("Top Teams by Goals")
plt.ylabel("Goals")
plt.xlabel("Team")
plt.show()
```

```python
top_ga = df.nlargest(10, "Goals_Assists")

sns.barplot(
    data=top_ga,
    x="Goals_Assists",
    y="Player_Name",
    palette="crest"
)

plt.title("Top 10 Goal Contributions")
plt.show()
```

```python
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
fig.suptitle('World Cup Player Statistics Analysis', fontsize=16, fontweight='bold')

# 1. Top 10 Goal Scorers
sns.barplot(data=df.nlargest(10, 'Goals'), x='Goals', y='Player_Name', ax=axes[0, 0], palette='viridis')
axes[0, 0].set_title('Top 10 Goal Scorers', fontweight='bold')
axes[0, 0].set_xlabel('Goals')
axes[0, 0].set_ylabel('')

# 2. Top 10 Assists
sns.barplot(data=df.nlargest(10, 'Assists'), x='Assists', y='Player_Name', ax=axes[0, 1], palette='plasma')
axes[0, 1].set_title('Top 10 Assist Providers', fontweight='bold')
axes[0, 1].set_xlabel('Assists')
axes[0, 1].set_ylabel('')

# 3. Top 10 Goal Contributions
sns.barplot(data=df.nlargest(10, 'Goals_Assists'), x='Goals_Assists', y='Player_Name', ax=axes[0, 2], palette='crest')
axes[0, 2].set_title('Top 10 Goal Contributions', fontweight='bold')
axes[0, 2].set_xlabel('Goals + Assists')
axes[0, 2].set_ylabel('')

# 4. Goals vs Assists Scatter
sns.scatterplot(data=df, x='Goals', y='Assists', hue='Position', ax=axes[1, 0], palette='Set2', s=50)
axes[1, 0].set_title('Goals vs Assists by Position', fontweight='bold')
axes[1, 0].legend(fontsize=8, loc='upper left')

# 5. Goals by Position
sns.boxplot(data=df, x='Position', y='Goals', ax=axes[1, 1], palette='Set3')
axes[1, 1].set_title('Goals Distribution by Position', fontweight='bold')
axes[1, 1].set_xlabel('Position')
axes[1, 1].set_ylabel('Goals')

# 6. Age Distribution
sns.histplot(data=df, x='Age', bins=15, kde=True, ax=axes[1, 2], color='skyblue')
axes[1, 2].set_title('Age Distribution', fontweight='bold')
axes[1, 2].set_xlabel('Age')
axes[1, 2].set_ylabel('Frequency')

# 7. Top Teams by Goals
team_goals_plot = df.groupby('Team')['Goals'].sum().sort_values(ascending=True).tail(10)
team_goals_plot.plot(kind='barh', ax=axes[2, 0], color='lightcoral')
axes[2, 0].set_title('Top 10 Teams by Goals', fontweight='bold')
axes[2, 0].set_xlabel('Total Goals')
axes[2, 0].set_ylabel('')

# 8. Yellow Cards Distribution
sns.histplot(data=df, x='Yellow_Cards', bins=10, kde=False, ax=axes[2, 1], color='khaki')
axes[2, 1].set_title('Yellow Cards Distribution', fontweight='bold')
axes[2, 1].set_xlabel('Yellow Cards')
axes[2, 1].set_ylabel('Frequency')

# 9. Correlation Heatmap (subset of most important correlations)
corr_subset = corr[['Goals', 'Assists', 'Goals_Assists', 'Minutes_Played', 'Age']].loc[['Goals', 'Assists', 'Goals_Assists', 'Minutes_Played', 'Age']]
sns.heatmap(corr_subset, annot=True, fmt='.2f', cmap='coolwarm', ax=axes[2, 2], cbar_kws={'shrink': 0.8})
axes[2, 2].set_title('Correlation Matrix', fontweight='bold')

plt.tight_layout()
plt.show()
```
