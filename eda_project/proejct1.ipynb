```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import os   
import kagglehub
```

```python
import pandas as pd
import os
import kagglehub

path = kagglehub.dataset_download("spscientist/students-performance-in-exams")
csv_file = os.path.join(path, "StudentsPerformance.csv")
df = pd.read_csv(csv_file)

```

```python
print(df.describe)
df.describe(include=object)
```

```python
print(df.duplicated().sum())
df.drop_duplicates(inplace=True)
```

```python
print(df.isnull().sum())
```

```python
print(df.dtypes)
```

```python
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_", regex=False)
)
print(df.columns)
```

```python
df['gender']=df['gender'].astype(object)
df['race/ethnicity']=df['race/ethnicity'].astype(object)
```

```python
check = {
    "Gender": df["gender"].unique(),
    "Race/Ethnicity": df["race/ethnicity"].unique(),
    "Parental Education": df["parental_level_of_education"].unique(),
    "lunch": df['lunch'].unique(),
    "test_prepration_course": df['test_preparation_course'].unique(),
}

print(check)
```

```python
# Box plots to visually spot outliers
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
df.boxplot(column='math_score', ax=axes[0])
df.boxplot(column='reading_score', ax=axes[1])
df.boxplot(column='writing_score', ax=axes[2])
plt.show()

# Also check min/max values
df[['math_score', 'reading_score', 'writing_score']].describe()
```

```python
df[['math_score','reading_score', 'writing_score']].hist(figsize= (12,4))
plt.show()
df.groupby('gender')[['math_score', 'reading_score', 'writing_score']].mean()
df.groupby('test_preparation_course')[['math_score']].mean()
sns.heatmap(df[['math_score', 'reading_score', 'writing_score']].corr(), annot=True)
plt.show()
```

```python
# Total score and average score
df['total_score'] = df['math_score'] + df['reading_score'] + df['writing_score']
df['average_score'] = df['total_score'] / 3
```

```python
df.to_csv('students_clean.csv', index=False)
```
