```python
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (16, 12)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Palmer Penguins Dataset', fontsize=20,fontweight='bold')

ax1=axes[0, 0]
sns.histplot(data=df, x='bill_length_mm', hue='species', kde=True, ax=ax1)
ax1.set_title('Distribution of Bill Length by Species', fontsize=16, fontweight='bold')  
ax1.set_xlabel('Bill Length (mm)', fontsize=14)
ax1.set_ylabel('Count', fontsize=14)

ax2=axes[1, 1]
sns.histplot(data=df, x='bill_depth_mm', hue='species', kde=True, ax=ax2)
ax2.set_title('Distribution of Bill Depth by Species', fontsize=16, fontweight='bold')
ax2.set_xlabel('Bill Depth (mm)', fontsize=14)
ax2.set_ylabel('Count', fontsize=14)

ax3 = axes[0, 1]
sns.lineplot(data=df, x='year', y='body_mass_g', hue='species', ax=ax3)
ax3.set_title('Body Mass over Years by Species', fontsize=16, fontweight='bold')
ax3.set_xlabel('Year', fontsize=14)
ax3.set_ylabel('Body Mass (g)', fontsize=14)

ax4 = axes[1, 0]
sns.barplot(data=df, x='species', y='flipper_length_mm', hue='sex', ax=ax4)
ax4.set_title('Flipper Length by Species and Sex', fontsize=16, fontweight='bold')
ax4.set_xlabel('Species', fontsize=14)
ax4.set_ylabel('Flipper Length (mm)', fontsize=14)
```
