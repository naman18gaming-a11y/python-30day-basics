```python
# Convert to datetime
df['Start_Date'] = pd.to_datetime(df['Start_Date'], errors='coerce', dayfirst=True)

df = df.dropna(subset=['Start_Date'])
# Drop rows that include time (keep only pure dates)
df = df[df['Start_Date'].dt.time == pd.to_datetime('00:00:00').time()]
```

```python
# ensure df stays a DataFrame, then fill missing Channel values
if isinstance(df, pd.Series):
    df = df.to_frame(name='Channel')
df['Channel'] = df['Channel'].fillna("UnKnown")
```

```python
df['Conversions'] = df['Conversions'].fillna(0)
```

```python
df['Spend'] = df['Spend'].astype(str).str.replace('$','', regex=False) 
df['Spend'] = df['Spend'].astype(str).apply(lambda x: '$' + x)

```

```python
df['Active'].value_counts()
```

```python
df['Active'] = (df['Active']
                .astype(str)
                .str.strip()
                .str.lower()
                .replace({'1': 'yes', 'true': 'yes', 'y': 'yes', 'yes': 'yes',
                          '0': 'no', 'false': 'no', 'n': 'no', 'no': 'no'}))
df['Active'] = df['Active'].map({'yes': 'Yes', 'no': 'No'}).fillna('No')
```

```python
df = df.drop(columns=['Clicks'])
```

```python
df.to_csv('market_clean_data.csv', index=False)
```
